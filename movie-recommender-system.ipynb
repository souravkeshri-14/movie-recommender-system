```python
import ast
```

```python
def convert(obj):
    list1 = []
    for i in ast.literal_eval(obj) :
        list1.append(i["name"])
    return list1
```

```python
def convert_cast(obj):
    list1 = []
    ounter = 0
    for i in ast.literal_eval(obj) :
        if counter < 3 :
            counter += 1
            list1.append(i["name"])
        else :
            return list1
```

```python
movie['cast'] = movie['cast'].apply(convert_cast)
```

```python
movies = movies.merge(credits["crew"],on="title")
```

```python
# First, check the column names in both DataFrames to identify the correct column names
print("Movies columns:", movies.columns.tolist())
print("Credits crew columns:", credits["crew"].columns.tolist())

# Option 1: If the column names are different, specify them explicitly
# Replace 'correct_movie_column' and 'correct_crew_column' with the actual column names
movies = movies.merge(credits["crew"], left_on="title", right_on="correct_crew_column")

# Option 2: If both DataFrames have the same column name but it's not 'title'
# Replace 'actual_column_name' with the correct column name found above
movies = movies.merge(credits["crew"], on="actual_column_name")

# Option 3: If you need to merge on index instead of a column
movies = movies.merge(credits["crew"], left_index=True, right_index=True)
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
```

```python
movies["crew"]=credits['crew'].apply(fetch_director)
```

```python
movies.dropna(inplace=True)
```

```python
movies.head()
```

```python
movies["overview"] =movies["overview"].apply(lambda x : x.split())
```

```python
movies.head()
```

```python
def remove_space(obj):
    l = []
    for i in obj :
        l.append(i.replace(" ",""))
    return l
```

```python
movies['cast'] = movies['cast'].apply(remove_space)
```

```python
movies['genres'] = movies['genres'].apply(remove_space)
```

```python
movies['keywords'] = movies['keywords'].apply(remove_space)
```

```python
movies['crew'] = movies['crew'].apply(remove_space)
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
new.head()
```

```python
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

```python
import nltk

from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new['tags'] = new['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector = cv.fit_transform(new['tags']).toarray()
```

```python
vector.shape
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new[new['title'] == 'The Lego Movie'].index[0]
```

```python
def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
```

```python
recommend('Spectre')
```
